## Project Overview
This project performs an end-to-end **Exploratory Data Analysis (EDA)** on a customer churn dataset.

The goal is to understand customer characteristics, identify patterns associated with churn, and generate business-oriented insights that can help a telecom company improve customer retention.

```python
def conv(value):
    if value==1:
        return "yes"
    else:
        return "no"

df['Senior Citizen']=df["Senior Citizen"].apply(conv)
```

COUNT OF CUTOMER BY CHURN


```python
ax=sns.countplot(x='Churn',data=df)

ax.bar_label(ax.containers[0])
plt.title("Count of customers by churn")
plt.show()
```

PERCENTAGE OF CHURN CUSTOMERS

```python
plt.figure(figsize=(3,4))
gb= df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb['Churn'],labels=gb.index,autopct="%1.2f%%")
plt.title("Percentage of churn customers",fontsize=10)
plt.show()
```

CHURN BY GENDER

```python
plt.figure(figsize=(3,3))
sns.countplot(x="gender",data=df,hue="Churn")
plt.title("Churn by gender")
plt.show()
```

COUNT OF CUSTOMER BY SENIOR CITIZEN

```python
plt.figure(figsize=(4,4))
ax=sns.countplot(x="Senior Citizen",data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of customers by Senior Citizen")
plt.show()
```

CHURN BY TENURE

```python
plt.figure(figsize=(9,4))
sns.histplot(x="tenure",data=df,bins=72,hue="Churn")
plt.show()
```

COUNT OF CUSTOMERS BY CONTRACT

```python
plt.figure(figsize=(4,4))
ax=sns.countplot(x= "Contract",data=df,hue="Churn")
ax.bar_label(ax.containers[0])
plt.title("Count of customers by contract")
plt.show()
```

CHURNED CUSTOMERS BY PAYMENT METHOD

```python
plt.figure(figsize=(6,4))
ax=sns.countplot(x="Payment Method",data=df,hue="Churn")
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.title("Churned customers by payment method")
plt.xticks(rotation=45)
plt.show()
```

Key Business Insights

Based on the EDA, the following patterns should be highlighted:

1. **Overall churn:** A measurable portion of the customer base has churned, making retention an important business concern.
2. **Contract type:** Month-to-month customers generally represent a higher-risk group than customers on longer contracts.
3. **Tenure:** Customers in the earliest tenure groups should be monitored carefully because early-stage customers can show higher churn risk.
4. **Billing:** Monthly charge differences between churned and retained customers can help identify pricing or value-perception patterns.
5. **Payment method:** Churn rates vary across payment methods, making billing experience an area worth investigating.
6. **Services:** Churn behavior differs across service categories, which can help guide customer support and service-bundle strategies.
7. **Demographics:** Gender and senior-citizen comparisons provide useful segmentation, but observed differences should not automatically be interpreted as causal.
8. **Correlation:** Numerical correlations can reveal associations, but they do not establish cause-and-effect relationships.

Conclusion

This exploratory analysis provides a structured view of customer churn across demographics, tenure, contracts, services, billing, and payment methods.

The analysis shows that churn is not evenly distributed across all customer groups. Contract type, tenure, charges, payment method, and service characteristics provide useful signals for understanding customer retention.
